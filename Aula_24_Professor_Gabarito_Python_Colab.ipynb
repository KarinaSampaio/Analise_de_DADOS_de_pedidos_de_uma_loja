```python
nome = "Ana"
idade = 28
salario = 4200.50
ativo = True

print(f"Nome: {nome} | Idade: {idade} | Salário: R$ {salario:.2f} | Ativo: {ativo}")
```

```python
valor = 235.0

if valor >= 200:
    desconto = 0.10
elif valor >= 100:
    desconto = 0.05
else:
    desconto = 0

valor_final = valor * (1 - desconto)
print(f"Valor final: R$ {valor_final:.2f}")
```

```python
vendas = [120.0, 80.5, 250.0, 99.9, 310.0]

total = 0
for venda in vendas:
    total += venda

print(f"Faturamento: R$ {total:.2f}")
```

```python
tentativas = ["123", "senha", "dados123"]
indice = 0
autenticado = False

while indice < len(tentativas) and not autenticado:
    if tentativas[indice] == "dados123":
        autenticado = True
    else:
        indice += 1

print("Acesso liberado!" if autenticado else "Acesso bloqueado.")
```

```python
nome = "Marina"
valor = 280.0

if valor >= 250:
    classificacao = "ALTO"
elif valor >= 100:
    classificacao = "MÉDIO"
else:
    classificacao = "BAIXO"

desconto = 0.10 if valor >= 200 else 0
valor_final = valor * (1 - desconto)

print(f"Cliente: {nome}")
print(f"Classificação: {classificacao}")
print(f"Desconto: {desconto*100:.0f}%")
print(f"Valor final: R$ {valor_final:.2f}")
```
